```python
from ultralytics import YOLO
import torch
from google.colab import drive

print(torch.cuda.is_available())
```

```python
model = YOLO('yolov8m.pt')

data_path = '/content/drive/MyDrive/chess_pieces_finetuned/data.yaml'

results = model.train(
    data=data_path,
    epochs=150,
    imgsz=512,
    project='finetuned_pieces_detection',
    name='run_real_medium',
    batch=8,
    patience=15
)
```
